### Демонстрация работы `parse_int`

```python
def parse_int(s: str) -> int | None:
    """Преобразует строку в int. Если это невозможно, возвращает None."""
    s = s.strip()  # убираем лишние пробелы по краям
    try:
        return int(s)
    except ValueError:
        return None
    
examples = ["12", " -7 ", "12a", ""]

for s in examples:
    result = parse_int(s)
    print(f"parse_int({s!r}) -> {result!r}")
```

## Часть 2. Циклы и работа со списками

Создадим список из 25 целых чисел, положительных и отрицательных.

```python
nums = [5, -3, 12, 1, 7, -8, 14, 21, -1, 9,
        4, -16, 33, 18, -5, 2, 11, -20, 6, 15,
        -9, 27, 8, -4, 10]

print("Список:", nums)
print("Количество элементов:", len(nums))
```

### `count_positive` — количество положительных чисел


```python
def count_positive(nums: list[int]) -> int:
    """Возвращает количество чисел больше 0."""
    count = 0
    for n in nums:
        if n > 0:
            count += 1
    return count


print("Положительных чисел:", count_positive(nums))
print("Проверка на [0, -1, 2]:", count_positive([0, -1, 2]))  # ноль не считается -> 1
print("Пустой список:", count_positive([]))
```

### `sum_even` — сумма чётных чисел

Число чётное, если остаток от деления на 2 равен нулю (`n % 2 == 0`). Отрицательные чётные числа и ноль тоже учитываются.

```python
def sum_even(nums: list[int]) -> int:
    """Возвращает сумму всех чётных чисел списка."""
    total = 0
    for n in nums:
        if n % 2 == 0:
            total += n
    return total


print("Сумма чётных чисел:", sum_even(nums))
print("Проверка на [1, 2, 3, 4]:", sum_even([1, 2, 3, 4]))  # 2 + 4 = 6
```

### `first_divisible` — первое число, делящееся на `k`


```python
def first_divisible(nums: list[int], k: int) -> int | None:
    """Возвращает первое число из списка, которое делится на k, или None."""
    if k == 0:
        raise ValueError("k не может быть равен 0")

    result = None
    for n in nums:
        if n % k == 0:
            result = n
            break  # нашли первое подходящее число и выходим из цикла
    return result


print("Первое, делящееся на 7:", first_divisible(nums, 7))
print("Первое, делящееся на 4:", first_divisible(nums, 4))
print("Первое, делящееся на 100:", first_divisible(nums, 100))

try:
    first_divisible(nums, 0)
except ValueError as e:
    print("Ошибка при k = 0:", e)
```

## Часть 3. Lambda и сортировки


```python
items = [
    {"name": "A", "price": 120, "rating": 4.7},
    {"name": "B", "price": 90, "rating": 4.2},
    {"name": "C", "price": 150, "rating": 4.9},
]
```

### Сортировка по `price` (по возрастанию)

```python
by_price = sorted(items, key=lambda item: item["price"])

for item in by_price:
    print(item)
```

### Сортировка по `rating` (по убыванию)

```python
by_rating = sorted(items, key=lambda item: item["rating"], reverse=True)

for item in by_rating:
    print(item)
```

Исходный список `items` при этом не изменился:

```python
print(items)
```

## Часть 4. Мини-ООП: класс `Chocolate`


```python
class Chocolate:
    def __init__(self, name: str, price: float):
        if price <= 0:
            raise ValueError(f"Цена должна быть больше 0, получено: {price}")
        self.name = name
        self.price = price

    def discount(self, percent: int) -> float:
        if not 0 <= percent <= 100:
            raise ValueError(f"Скидка должна быть от 0 до 100, получено: {percent}")
        return round(self.price * (1 - percent / 100), 2)

    def __repr__(self) -> str:
        return f"Chocolate(name={self.name!r}, price={self.price})"
```

### Создание объектов и вызов `discount`

```python
milka = Chocolate("Milka", 120.0)
alpen_gold = Chocolate("Alpen Gold", 89.9)

print(milka)
print(alpen_gold)

print(f"{milka.name}: скидка 10% -> {milka.discount(10)}")
print(f"{alpen_gold.name}: скидка 25% -> {alpen_gold.discount(25)}")

# Поле price не изменилось
print(f"Цена {milka.name} после вызова discount:", milka.price)
```

### Обработка ошибок

```python
for bad_price in [0, -50]:
    try:
        Chocolate("Bad", bad_price)
    except ValueError as e:
        print("Ошибка:", e)

try:
    milka.discount(150)
except ValueError as e:
    print("Ошибка:", e)
```
